#Preparación de datos

```python
# Standardize column names (optional, but good practice)
df_defunciones.columns = df_defunciones.columns.str.strip().str.replace(' ', '_')
df_poblacion.columns = df_poblacion.columns.str.strip().str.replace(' ', '_').str.replace('(1=Hombre_2=Mujer)', '', regex=False)

# Rename the 'Sexo_' column to 'Sexo' in df_poblacion
df_poblacion.rename(columns={'Sexo_': 'Sexo'}, inplace=True)

# Rename 'AÑO' in df_defunciones to 'Año' to match df_poblacion
df_defunciones.rename(columns={'AÑO': 'Año'}, inplace=True)

# Convert 'Edad' in df_defunciones to integer
df_defunciones['Edad'] = df_defunciones['Edad'].astype(int)

# Convert 'Sexo' in df_defunciones to integer
df_defunciones['Sexo'] = df_defunciones['Sexo'].astype(int)

# --- Age Grouping for Cancer Analysis ---
# Define new age group bins and labels for the desired grouping (0-19, 20-59, 60+)
new_age_bins = [-1, 19, 59, float('inf')]
new_age_labels = ['0-19', '20-59', '60+']

# The 'Edad' column in df_poblacion is already in integer format from the parquet file,
# so direct mapping from 'Grupo_edad' string is not necessary for creating the new age groups with pd.cut.
# If you need to work with the original 'Grupo_edad' strings for other purposes, keep the mapping code.
# Otherwise, this cell will now focus on preparing data for the new age grouping in subsequent cells.


# Convert 'Año' in df_poblacion to integer
df_poblacion['Año'] = df_poblacion['Año'].astype(int)


# Display data types and head to confirm changes
display(df_defunciones.info())
display(df_poblacion.info())

display(df_defunciones.head())
display(df_poblacion.head())
```

#Calculo de tasas específicas y ajustadas por edad de la población

```python
# Filter for Cancer deaths with grouped ages
df_cancer_defunciones_grouped = df_defunciones[df_defunciones['CAPITULO_DIAG1'] == 'C00-D48'].copy()

# Filter for Gastric Cancer deaths with grouped ages
df_gastric_cancer_defunciones_grouped = df_defunciones[df_defunciones['CODIGO_CATEGORIA_DIAG1'] == 'C16'].copy()

# Define new age group bins and labels based on the requested grouping
# Group 0-19, 20-59, and 60+
new_age_bins = [-1, 19, 59, float('inf')]
new_age_labels = ['0-19', '20-59', '60+']


# Apply the new age grouping to the death dataframes using the 'Edad' column
df_cancer_defunciones_grouped['Edad_Grupo_Cancer'] = pd.cut(
    df_cancer_defunciones_grouped['Edad'],
    bins=new_age_bins,
    labels=new_age_labels,
    right=True,
    include_lowest=True
)

df_gastric_cancer_defunciones_grouped['Edad_Grupo_Cancer'] = pd.cut(
    df_gastric_cancer_defunciones_grouped['Edad'],
    bins=new_age_bins,
    labels=new_age_labels,
    right=True,
    include_lowest=True
)

# --- Age Grouping and Aggregation for Population Data ---
# Define a mapping for age groups to a numerical representation (e.g., lower bound of age group)
age_group_mapping = {
    '0 Año': 0,
    'De 1 a 4 años': 1,
    'De 5 a 9 años': 5,
    'De 10 a 14 años': 10,
    'De 15 a 19 años': 15,
    'De 20 a 24 años': 20,
    'De 25 a 29 años': 25,
    'De 30 a 34 años': 30,
    'De 35 a 39 años': 35,
    'De 40 a 44 años': 40,
    'De 45 a 49 años': 45,
    'De 50 a 54 años': 50,
    'De 55 a 59 años': 55,
    'De 60 a 64 años': 60,
    'De 65 a 69 años': 65,
    'De 70 a 74 años': 70,
    'De 75 a 79 años': 75,
    '80 años y más': 80
}

# Apply the mapping to create a new numerical age column
df_poblacion['Edad_Numerical_Approx'] = df_poblacion['Grupo_edad'].map(age_group_mapping)

# Apply the new age grouping to the population dataframe using the numerical approximation and new_age_labels
df_poblacion['Edad_Grupo_Cancer'] = pd.cut(
    df_poblacion['Edad_Numerical_Approx'],
    bins=new_age_bins,
    labels=new_age_labels,
    right=True,
    include_lowest=True
)

# Filter out any rows where the age group mapping resulted in NaN (if any)
df_poblacion_cleaned = df_poblacion.dropna(subset=['Edad_Grupo_Cancer']).copy()


# Aggregate population by Año, Comuna, Sexo, and Edad_Grupo_Cancer
df_poblacion_agg_cancer_grouped = df_poblacion_cleaned.groupby(['Año', 'Comuna', 'Sexo', 'Edad_Grupo_Cancer'], observed=True)['Poblacion'].sum().reset_index()


# Aggregate deaths for Cancer by Año, Comuna, Sexo, and Edad_Grupo_Cancer
df_cancer_deaths_agg_cancer_grouped = df_cancer_defunciones_grouped.groupby(['Año', 'Comuna', 'Sexo', 'Edad_Grupo_Cancer'], observed=True).size().reset_index(name='Cancer_Deaths')

# Aggregate deaths for Gastric Cancer by Año, Comuna, Sexo, and Edad_Grupo_Cancer
df_gastric_cancer_deaths_agg_cancer_grouped = df_gastric_cancer_defunciones_grouped.groupby(['Año', 'Comuna', 'Sexo', 'Edad_Grupo_Cancer'], observed=True).size().reset_index(name='Gastric_Cancer_Deaths')


# Merge aggregated cancer deaths with new grouped population data
df_cancer_rates_age_specific_cancer_grouped = pd.merge(
    df_cancer_deaths_agg_cancer_grouped,
    df_poblacion_agg_cancer_grouped,
    on=['Año', 'Comuna', 'Sexo', 'Edad_Grupo_Cancer'],
    how='left'
)

# Merge aggregated gastric cancer deaths with new grouped population data, using left merge from population to ensure all population groups are included
df_gastric_cancer_rates_age_specific_cancer_grouped = pd.merge(
    df_poblacion_agg_cancer_grouped,
    df_gastric_cancer_deaths_agg_cancer_grouped,
    on=['Año', 'Comuna', 'Sexo', 'Edad_Grupo_Cancer'],
    how='left'
)


# Fill NaN values in Gastric_Cancer_Deaths with 0 after the merge
df_gastric_cancer_rates_age_specific_cancer_grouped['Gastric_Cancer_Deaths'] = df_gastric_cancer_rates_age_specific_cancer_grouped['Gastric_Cancer_Deaths'].fillna(0)

# Fill NaN population values with 0
df_cancer_rates_age_specific_cancer_grouped['Poblacion'] = df_cancer_rates_age_specific_cancer_grouped['Poblacion'].fillna(0)
df_gastric_cancer_rates_age_specific_cancer_grouped['Poblacion'] = df_gastric_cancer_rates_age_specific_cancer_grouped['Poblacion'].fillna(0)


# Calculate age-specific mortality rates for Cancer with new grouped ages
df_cancer_rates_age_specific_cancer_grouped['Cancer_Age_Specific_Rate'] = (
    df_cancer_rates_age_specific_cancer_grouped['Cancer_Deaths'] / df_cancer_rates_age_specific_cancer_grouped['Poblacion']
) * 100000

# Calculate age-specific mortality rates for Gastric Cancer with new grouped ages
df_gastric_cancer_rates_age_specific_cancer_grouped['Gastric_Cancer_Age_Specific_Rate'] = (
    df_gastric_cancer_rates_age_specific_cancer_grouped['Gastric_Cancer_Deaths'] / df_gastric_cancer_rates_age_specific_cancer_grouped['Poblacion']
) * 100000

# Replace infinite values with 0 in the rate columns
import numpy as np
df_cancer_rates_age_specific_cancer_grouped.replace([np.inf, -np.inf], 0, inplace=True)
df_gastric_cancer_rates_age_specific_cancer_grouped.replace([np.inf, -np.inf], 0, inplace=True)

# Fill NaN values in the rate columns with 0 after calculation
df_gastric_cancer_rates_age_specific_cancer_grouped['Gastric_Cancer_Age_Specific_Rate'] = df_gastric_cancer_rates_age_specific_cancer_grouped['Gastric_Cancer_Age_Specific_Rate'].fillna(0)


# --- Calculate Age-Adjusted Rates using National Population by new age groups as the standard ---

# Define the 2017 census population for the standard population by the new age groups
standard_population_2017 = {
    '0-19': 4768447,
    '20-59': 9955385,
    '60+': 2850171
}

# Create a DataFrame for the standard population distribution
df_standard_pop_dist = pd.DataFrame(list(standard_population_2017.items()), columns=['Edad_Grupo_Cancer', 'Standard_Population'])


# Calculate the total size of the standard population
total_standard_population = df_standard_pop_dist['Standard_Population'].sum()


# Merge age-specific cancer rates with the standard population distribution
df_cancer_rates_standardized = pd.merge(
    df_cancer_rates_age_specific_cancer_grouped,
    df_standard_pop_dist,
    on='Edad_Grupo_Cancer',
    how='left'
)

df_gastric_cancer_rates_standardized = pd.merge(
    df_gastric_cancer_rates_age_specific_cancer_grouped,
    df_standard_pop_dist,
    on='Edad_Grupo_Cancer',
    how='left'
)

# Calculate expected deaths in the standard population for each group
df_cancer_rates_standardized['Expected_Deaths_Standard_Pop'] = (
    df_cancer_rates_standardized['Cancer_Age_Specific_Rate'] / 100000
) * df_cancer_rates_standardized['Standard_Population']

df_gastric_cancer_rates_standardized['Expected_Deaths_Standard_Pop'] = (
    df_gastric_cancer_rates_standardized['Gastric_Cancer_Age_Specific_Rate'] / 100000
) * df_gastric_cancer_rates_standardized['Standard_Population']

# Fill NaN expected deaths with 0 after calculation
df_gastric_cancer_rates_standardized['Expected_Deaths_Standard_Pop'] = df_gastric_cancer_rates_standardized['Expected_Deaths_Standard_Pop'].fillna(0)


# Aggregate expected deaths by Año, Comuna, and Sexo to get total expected deaths in standard population
df_cancer_age_adjusted_rates = df_cancer_rates_standardized.groupby(['Año', 'Comuna', 'Sexo'])['Expected_Deaths_Standard_Pop'].sum().reset_index(name='Total_Expected_Deaths_Standard_Pop')
df_gastric_cancer_age_adjusted_rates = df_gastric_cancer_rates_standardized.groupby(['Año', 'Comuna', 'Sexo'])['Expected_Deaths_Standard_Pop'].sum().reset_index(name='Total_Expected_Deaths_Standard_Pop')


# Calculate the age-adjusted rate
# Ensure we are dividing by the total standard population
df_cancer_age_adjusted_rates['Cancer_Age_Adjusted_Rate'] = (
    df_cancer_age_adjusted_rates['Total_Expected_Deaths_Standard_Pop'] / total_standard_population
) * 100000

df_gastric_cancer_age_adjusted_rates['Gastric_Cancer_Age_Adjusted_Rate'] = (
    df_gastric_cancer_age_adjusted_rates['Total_Expected_Deaths_Standard_Pop'] / total_standard_population
) * 100000

# Fill NaN age-adjusted rates with 0 after calculation
df_gastric_cancer_age_adjusted_rates['Gastric_Cancer_Age_Adjusted_Rate'] = df_gastric_cancer_age_adjusted_rates['Gastric_Cancer_Age_Adjusted_Rate'].fillna(0)


# Display the resulting age-adjusted rates
display(df_cancer_age_adjusted_rates.head())
display(df_gastric_cancer_age_adjusted_rates.head())

# Display age-specific rates for Comuna 1405, Año 2002
print("\nAge-Specific Rates for Comuna 1405, Año 2002 (All Cancer):")
display(df_cancer_rates_age_specific_cancer_grouped[(df_cancer_rates_age_specific_cancer_grouped['Comuna'] == 1405) & (df_cancer_rates_age_specific_cancer_grouped['Año'] == 2002)])

print("\nAge-Specific Rates for Comuna 1405, Año 2002 (Gastric Cancer):")
display(df_gastric_cancer_rates_age_specific_cancer_grouped[(df_gastric_cancer_rates_age_specific_cancer_grouped['Comuna'] == 1405) & (df_gastric_cancer_rates_age_specific_cancer_grouped['Año'] == 2002)])
```

# Summary of Calculated Mortality Rates and DataFrames

Here is a summary of the different mortality rates calculated and the pandas DataFrames where they are stored:

    *   Contains the age-adjusted mortality rate for all cancers, allowing comparison across years and communes while accounting for age differences, broken down by sex.

2.  **Age-Adjusted Rates for Gastric Cancer (by Year, Commune, and Sex):**
    *   DataFrame: `df_gastric_cancer_age_adjusted_rates`

```python
# Save the intermediate dataframes used for final combinations to parquet files
output_dir = path_tasas

# Dataframes used in cell 87f42b7c to create df_combined_final:
# df_cancer_rates_age_specific_cancer_grouped
# df_gastric_cancer_rates_age_specific_cancer_grouped
# df_cancer_rates_standardized
# df_gastric_cancer_rates_standardized
# df_poblacion_agg_cancer_grouped (used to calculate total population)
# df_crude_rates_calculated (calculated using intermediate dataframes and used in the final merge)
# df_cancer_age_adjusted_rates
# df_gastric_cancer_age_adjusted_rates


# Dataframes used in cell a4af0c5a to create df_combined_final_no_sex:
# df_cancer_rates_age_specific_cancer_grouped_no_sex
# df_gastric_cancer_rates_age_specific_cancer_grouped_no_sex
# df_cancer_rates_standardized_no_sex
# df_gastric_cancer_rates_standardized_no_sex
# df_crude_rates_calculated (already listed above)
# df_cancer_age_adjusted_rates_no_sex
# df_gastric_cancer_age_adjusted_rates_no_sex
# df_poblacion_agg_cancer_grouped_no_sex (used to calculate total population, derived from df_poblacion_cleaned in 6d061582 which is aggregated in 6d061582)

# Calculate total population by Año and Comuna
df_total_population_comuna_year = df_poblacion_agg_cancer_grouped.groupby(['Año', 'Comuna'])['Poblacion'].sum().reset_index(name='Total_Population')

# Calculate total deaths for all cancer by Año and Comuna
df_cancer_total_deaths_comuna_year = df_cancer_defunciones_grouped.groupby(['Año', 'Comuna']).size().reset_index(name='Total_Cancer_Deaths')

# Calculate total deaths for gastric cancer by Año and Comuna
df_gastric_cancer_total_deaths_comuna_year = df_gastric_cancer_defunciones_grouped.groupby(['Año', 'Comuna']).size().reset_index(name='Total_Gastric_Cancer_Deaths')


# Merge total population and total deaths for crude rate calculation
df_crude_rates_calculated = pd.merge(
    df_total_population_comuna_year,
    df_cancer_total_deaths_comuna_year,
    on=['Año', 'Comuna'],
    how='left'
)

df_crude_rates_calculated = pd.merge(
    df_crude_rates_calculated,
    df_gastric_cancer_total_deaths_comuna_year,
    on=['Año', 'Comuna'],
    how='left'
)

# Fill NaN death values with 0
df_crude_rates_calculated['Total_Cancer_Deaths'] = df_crude_rates_calculated['Total_Cancer_Deaths'].fillna(0)
df_crude_rates_calculated['Total_Gastric_Cancer_Deaths'] = df_crude_rates_calculated['Total_Gastric_Cancer_Deaths'].fillna(0)


# Calculate crude rate for all cancer
df_crude_rates_calculated['Cancer_Crude_Rate'] = (
    df_crude_rates_calculated['Total_Cancer_Deaths'] / df_crude_rates_calculated['Total_Population']
) * 100000

# Calculate crude rate for gastric cancer
df_crude_rates_calculated['Gastric_Cancer_Crude_Rate'] = (
    df_crude_rates_calculated['Total_Gastric_Cancer_Deaths'] / df_crude_rates_calculated['Total_Population']
) * 100000

# Replace infinite values with 0 in the calculated crude rate columns
import numpy as np
df_crude_rates_calculated.replace([np.inf, -np.inf], 0, inplace=True)

# Saving the relevant intermediate dataframes:
df_cancer_rates_age_specific_cancer_grouped.to_parquet(f'{output_dir}Cancer_Age_Specific_Rates_Grouped.parquet', index=False)
df_gastric_cancer_rates_age_specific_cancer_grouped.to_parquet(f'{output_dir}Gastric_Cancer_Age_Specific_Rates_Grouped.parquet', index=False)
df_cancer_rates_standardized.to_parquet(f'{output_dir}Cancer_Rates_Standardized.parquet', index=False)
df_gastric_cancer_rates_standardized.to_parquet(f'{output_dir}Gastric_Cancer_Rates_Standardized.parquet', index=False)
df_poblacion_agg_cancer_grouped.to_parquet(f'{output_dir}Population_Agg_Cancer_Grouped.parquet', index=False)
df_crude_rates_calculated.to_parquet(f'{output_dir}Crude_Rates_Calculated.parquet', index=False)
df_cancer_age_adjusted_rates.to_parquet(f'{output_dir}Cancer_Age_Adjusted_Rates.parquet', index=False)
df_gastric_cancer_age_adjusted_rates.to_parquet(f'{output_dir}Gastric_Cancer_Age_Adjusted_Rates.parquet', index=False)
df_cancer_rates_age_specific_cancer_grouped_no_sex.to_parquet(f'{output_dir}Cancer_Age_Specific_Rates_Grouped_No_Sex.parquet', index=False)
df_gastric_cancer_rates_age_specific_cancer_grouped_no_sex.to_parquet(f'{output_dir}Gastric_Cancer_Age_Specific_Rates_Grouped_No_Sex.parquet', index=False)
df_cancer_rates_standardized_no_sex.to_parquet(f'{output_dir}Cancer_Rates_Standardized_No_Sex.parquet', index=False)
df_gastric_cancer_rates_standardized_no_sex.to_parquet(f'{output_dir}Gastric_Cancer_Rates_Standardized_No_Sex.parquet', index=False)
df_cancer_age_adjusted_rates_no_sex.to_parquet(f'{output_dir}Cancer_Age_Adjusted_Rates_No_Sex.parquet', index=False)
df_gastric_cancer_age_adjusted_rates_no_sex.to_parquet(f'{output_dir}Gastric_Cancer_Age_Adjusted_Rates_No_Sex.parquet', index=False)
df_poblacion_agg_cancer_grouped_no_sex.to_parquet(f'{output_dir}Population_Agg_Cancer_Grouped_No_Sex.parquet', index=False)


print(f"Saved intermediate dataframes to {output_dir}")
```
